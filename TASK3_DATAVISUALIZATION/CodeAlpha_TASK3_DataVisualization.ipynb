```python
# CodeAlpha Task 3 - Data Visualization

## Project Overview

#This project is part of the CodeAlpha Data Analytics Internship.

#The objective is to transform the Sample Superstore dataset into meaningful
visualizations that help understand sales and profit performance.

## Visualizations Created

- Total Sales by Category
- Total Profit by Category
- Total Sales by Region
- Total Profit by Region
- Total Sales by Sub-Category
- Total Profit by Sub-Category

## Tools Used

- Python
- Pandas
- Matplotlib
- Google Colab

## Conclusion

The visualizations provide a clear comparison of sales and profit across
different categories, regions, and sub-categories.
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("SampleSuperstore.csv")

df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(category_sales.index, category_sales.values)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(category_profit.index, category_profit.values)

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(region_sales.index, region_sales.values)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(region_profit.index, region_profit.values)

plt.title("Total Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")

plt.show()
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(subcategory_sales.index, subcategory_sales.values)

plt.title("Total Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.show()
```

```python
subcategory_profit = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(subcategory_profit.index, subcategory_profit.values)

plt.title("Total Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)

plt.show()
```
